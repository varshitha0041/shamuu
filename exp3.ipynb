```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data={
    'number':[1,2,3,4,5],
    'pencil':[300,350,400,500,520],
    'textbook':[250,350,400,420,500],
    'drawingsheet':[100,200,200,250,300],
    'totalunit':[800,1000,1320,1510,2000],
    'profit':[8000,9500,10256,12000,18000],
  }
df=pd.DataFrame(data)
print(df)
df.info()
plt.plot(df['number'],df['profit'])
plt.xlabel("number")
plt.ylabel("totalprofit")
plt.show()
print("missing value")
print(df.isnull().sum())
print("sum of total profit")
print(df["profit"].sum())
print("max value of drawingsheet column")
print(df["drawingsheet"].max())
print("min value of drawingsheet column")
print(df["drawingsheet"].min())
new_unit=df['totalunit']+5
print(new_unit)
```
