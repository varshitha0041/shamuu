```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
cars_df=pd.read_csv('/content/mtcars (11).csv')
print(cars_df)
a=plt.scatter(cars_df['mpg'],cars_df['wt'],marker='*')
print(a)
plt.xlabel('mpg')
plt.ylabel('wieght')
plt.title('relationship between wieght of the cars and mpg')
plt.show()
trans=cars_df['am'].value_counts()
print('trans')
trans.plot.bar()
plt.xlabel('transmission')
plt.title('bar plot to check the frequence distribution of transmission type')
plt.show()
cars_df['mpg'].plot(kind='box')
plt.xlabel('mpg')
plt.title('box and whisker plot of mpg')
plt.show()
min_value=np.min(cars_df['mpg'])
q1=np.percentile(cars_df['mpg'],25)
median=np.median(cars_df['mpg'])
q3=np.percentile(cars_df['mpg'],75)
max_value=np.max(cars_df['mpg'])
print('minimum value:',min_value)
print('Q1:',q1)
print('median:',median)
print('Q3:',q3)
print('maximum valu:',max_value)
```
